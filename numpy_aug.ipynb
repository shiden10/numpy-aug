```python
#create a 4x4 matrix with float values

z=np.array([[3.1,4.5,7.7,3.5],[2.8,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z)

print(np.floor(z))
print(np.ceil(z))
print(np.round(z)) #rounding the closest even values
print(z.sum())
print(z.min())
print(z.max())
print(z.max(axis=1))
print(z.max(axis=0))
print(z.min(axis=1))
print(z.min(axis=0))
print(z.mean(axis=1))
print(z.mean(axis=0))
print(z.std(axis=1))
print(z.std(axis=0))
print(np.sqrt(z))
print(np.sin(z))
print(np.tan(z))
print(np.sort(z)) # to sort thematrix in a row wise method
print(np.argsort(z,axis=1)) #to sort the indexing values based on its pare


```
